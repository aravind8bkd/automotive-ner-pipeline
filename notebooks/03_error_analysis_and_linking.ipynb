# Error analysis and entity linking

Review false negatives and false positives against the labeled fixture, then propose taxonomy updates. Confirm an alias does not create ambiguity before merging a taxonomy change.

```python
import json
from pathlib import Path

from automotive_ner.extractor import Extractor

rows = [json.loads(line) for line in Path('../data/sample_reviews.jsonl').read_text(encoding='utf-8').splitlines()]
extractor = Extractor()
for row in rows:
    _, predictions = extractor.extract(row['text'])
    expected = {entity['canonical_id'] for entity in row.get('entities', [])}
    predicted = {entity['canonical_id'] for entity in predictions}
    if expected != predicted:
        print(row['review_id'], 'FN:', expected - predicted, 'FP:', predicted - expected, row['text'])
```

Entity offsets returned by the extractor index `redacted_text`. The linker intentionally maps aliases to one canonical ID; it does not infer a component from context or resolve ambiguous mentions.
